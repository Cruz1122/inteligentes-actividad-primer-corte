# Punto 2: Juego de las piedras con 1, 2 o 4

Quien toma la última piedra gana. En cada turno se pueden retirar `1`, `2` o `4` piedras, no 3. El estado es la cantidad de piedras. MAX mueve primero.

```python
MOVIMIENTOS = (1, 2, 4)


def movimientos_validos(piedras):
    return [m for m in MOVIMIENTOS if m <= piedras]


for n in range(1, 8):
    print(n, "piedras ->", movimientos_validos(n))
```

```python
def minimax_piedras(piedras, turno_max):
    # Si no quedan piedras, el jugador anterior tomó la última y ganó.
    if piedras == 0:
        return -1 if turno_max else 1

    valores = []
    for retirar in movimientos_validos(piedras):
        valor = minimax_piedras(piedras - retirar, not turno_max)
        valores.append(valor)

    return max(valores) if turno_max else min(valores)


def mejor_movimiento_piedras(piedras):
    opciones = []
    for retirar in movimientos_validos(piedras):
        valor = minimax_piedras(piedras - retirar, False)
        opciones.append((valor, retirar))

    mejor_valor, mejor_movimiento = max(opciones, key=lambda item: item[0])
    return {
        "retirar": mejor_movimiento,
        "valor": mejor_valor,
        "opciones": opciones,
    }
```

```python
resultados_piedras = {}
for piedras in range(1, 16):
    resultados_piedras[piedras] = mejor_movimiento_piedras(piedras)

resultados_piedras
```

```python
ganadoras = [n for n, r in resultados_piedras.items() if r["valor"] == 1]
perdedoras = [n for n, r in resultados_piedras.items() if r["valor"] == -1]

{
    "ganadoras": ganadoras,
    "perdedoras": perdedoras,
}
```
