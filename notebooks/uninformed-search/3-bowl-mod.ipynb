# Punto 3: Recipientes con capacidades diferentes

```python
from collections import deque
```

## Problema: recipientes de agua

Estado: `(m, n)`, donde `m` y `n` son las cantidades actuales de agua en cada recipiente.


```python
def sucesores_recipientes(estado, M, N):
    # Cantidad actual de agua en cada recipiente.
    m, n = estado

    # Cada elemento contiene:
    # (nuevo_estado, descripción_de_la_acción)
    return [
        # Llenar completamente M.
        ((M, n), "Llenar M"),

        # Llenar completamente N.
        ((m, N), "Llenar N"),

        # Vaciar completamente M.
        ((0, n), "Vaciar M"),

        # Vaciar completamente N.
        ((m, 0), "Vaciar N"),

        # Verter N -> M.
        # M recibe tanta agua como pueda sin superar su capacidad.
        # N conserva únicamente el agua que no cupo en M.
        (
            (
                min(M, m + n),
                max(0, m + n - M)
            ),
            "Verter N → M"
        ),

        # Verter M -> N.
        # N recibe tanta agua como pueda sin superar su capacidad.
        # M conserva únicamente el agua que no cupo en N.
        (
            (
                max(0, m + n - N),
                min(N, m + n)
            ),
            "Verter M → N"
        ),
    ]
```

```python
def bfs_recipientes(M, N, objetivo):
    # Ambos recipientes empiezan vacíos.
    inicio = (0, 0)

    # Cada elemento contiene:
    # (estado_actual, lista_de_estados, lista_de_acciones)
    cola = deque([(inicio, [inicio], [])])

    # Estados que ya fueron descubiertos.
    visitados = {inicio}

    while cola:
        # BFS toma siempre el estado más antiguo.
        estado, camino, acciones = cola.popleft()

        # El objetivo se cumple si alguno de los recipientes
        # contiene exactamente la cantidad deseada.
        if objetivo in estado:
            return {
                "estados": camino,
                "acciones": acciones,
                "explorados": len(visitados),
            }

        # Generamos todas las operaciones posibles desde el estado actual.
        for nuevo_estado, accion in sucesores_recipientes(estado, M, N):
            # Evitamos repetir estados.
            if nuevo_estado not in visitados:
                visitados.add(nuevo_estado)

                # Guardamos el nuevo estado junto con el historial necesario
                # para reconstruir la solución completa.
                cola.append(
                    (
                        nuevo_estado,
                        camino + [nuevo_estado],
                        acciones + [accion],
                    )
                )

    # No se encontró una configuración que cumpla el objetivo.
    return None


# Baseline original: recipientes de 5 y 3 litros, buscando exactamente 4 litros.
# solucion_recipientes = bfs_recipientes(5, 3, 4)
solucion_recipientes = bfs_recipientes(11, 5, 8)
solucion_recipientes
```

```python
def mostrar_solucion_recipientes(solucion):
    # Caso en el que BFS no encontró una solución.
    if solucion is None:
        print("No se encontró solución.")
        return

    # Extraemos los estados y acciones encontrados por BFS.
    estados = solucion["estados"]
    acciones = solucion["acciones"]

    print("Estado inicial:", estados[0])

    # Cada acción lleva al estado ubicado en la siguiente posición.
    for i, accion in enumerate(acciones, start=1):
        print(f"{i}. {accion:15s} -> {estados[i]}")

    print("\nEstados explorados:", solucion["explorados"])


mostrar_solucion_recipientes(solucion_recipientes)
```
